### 🤔 2. Was sind Embeddings?

Stell dir vor, du möchtest einem Computer beibringen, die Bedeutung von Wörtern zu verstehen. Computer verstehen aber nur Zahlen. Hier kommen Embeddings ins Spiel.

**Analogie:**
Denk an eine Bibliothek. Bücher über ähnliche Themen stehen oft nahe beieinander in den Regalen. Embeddings versuchen, ein ähnliches "Regalsystem" für Wörter oder Sätze zu erstellen, aber in einem mathematischen Raum.
* "König" und "Königin" wären nah beieinander.
* "Apfel" und "Banane" wären ebenfalls nah beieinander.
* "König" und "Banane" wären weiter voneinander entfernt.

**Technische Definition:**
Ein Embedding ist ein Vektor (eine Liste von Zahlen) fester Länge, der ein Objekt (z.B. ein Wort, einen Satz) repräsentiert. Die Werte in diesem Vektor werden so gelernt, dass semantisch ähnliche Objekte auch mathematisch ähnliche Vektoren haben (z.B. gemessen durch die Kosinus-Ähnlichkeit).

**Dimension des Vektorraums:**
Die "Dimension" eines Embeddings bezieht sich auf die Anzahl der Zahlen im Vektor. Gängige Dimensionen sind z.B. 384, 768 oder 1024. Eine höhere Dimension kann mehr Nuancen erfassen, erfordert aber auch mehr Rechenleistung.

**Warum sind Embeddings nützlich?**
* **Semantische Suche:** Finde Dokumente oder Sätze, die inhaltlich ähnlich zu einer Suchanfrage sind, auch wenn sie nicht genau dieselben Wörter enthalten.
* **Textklassifikation:** Bestimme die Kategorie eines Textes (z.B. Spam/Nicht-Spam, positiv/negativ).
* **Clustering:** Gruppiere ähnliche Texte.
* **Empfehlungssysteme:** Empfehle Nutzern Artikel oder Produkte basierend auf Ähnlichkeiten.
* **Frage-Antwort-Systeme:** Verstehe die Bedeutung von Fragen und finde passende Antworten.

### � 3. Embeddings generieren mit `sentence-transformers` (Hugging Face)

`sentence-transformers` ist eine fantastische Bibliothek, die den Zugriff auf viele vortrainierte Modelle (oft von Hugging Face) sehr einfach macht. Diese Modelle können Sätze oder Absätze in aussagekräftige Vektoren umwandeln.

Wir verwenden hier ein populäres, mehrsprachiges Modell: `paraphrase-multilingual-MiniLM-L12-v2`. Es ist relativ klein und effizient, aber liefert gute Ergebnisse für viele Aufgaben und benötigt keinen API-Key für die Nutzung.

````python
from sentence_transformers import SentenceTransformer
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Modell laden
# Beim ersten Mal wird das Modell heruntergeladen. Das kann einen Moment dauern.
# Dieses Modell ist öffentlich und erfordert keinen API-Key.
model_name_hf = 'paraphrase-multilingual-MiniLM-L12-v2'
model_hf = SentenceTransformer(model_name_hf)

print(f"Modell '{model_name_hf}' geladen.")

# Beispieltexte
sentences_hf = [
    "Das Wetter heute ist sonnig und warm.",
    "Ich genieße den Sonnenschein und die Wärme.",
    "Morgen soll es regnen.",
    "Künstliche Intelligenz verändert die Welt.",
    "Ein Apfel am Tag hält den Doktor fern."
]

# Embeddings generieren
embeddings_hf = model_hf.encode(sentences_hf)

# Die Form der Embeddings anzeigen
# (Anzahl der Sätze, Dimension der Embeddings)
print(f"\nForm der Embedding-Matrix (Hugging Face): {embeddings_hf.shape}")
print(f"Jeder Satz wurde in einen Vektor mit {embeddings_hf.shape[1]} Dimensionen umgewandelt.")

# Beispiel-Embedding für den ersten Satz
print(f"\nEmbedding für den ersten Satz ('{sentences_hf[0]}'):")
print(embeddings_hf[0][:10]) # Nur die ersten 10 Dimensionen anzeigen
````

**Was ist hier passiert?**
1.  Wir haben das `SentenceTransformer`-Modell geladen.
2.  Wir haben eine Liste von Beispielsätzen definiert.
3.  Die Methode `model_hf.encode()` hat jeden Satz in einen numerischen Vektor (Embedding) umgewandelt.
4.  Das Ergebnis `embeddings_hf` ist eine NumPy-Matrix, bei der jede Zeile das Embedding eines Satzes darstellt. Die Anzahl der Spalten ist die Dimension des Embedding-Vektors (für dieses Modell 384).

### 📏 4. Anwendungsbeispiel: Semantische Ähnlichkeit

Einer der häufigsten Anwendungsfälle für Embeddings ist die Berechnung der semantischen Ähnlichkeit zwischen Texten. Wenn zwei Texte ähnliche Embeddings haben, sind sie wahrscheinlich auch inhaltlich ähnlich.

Die **Kosinus-Ähnlichkeit** ist ein gängiges Maß, um die Ähnlichkeit zwischen zwei Vektoren zu berechnen. Sie misst den Kosinus des Winkels zwischen den Vektoren.
* Ein Wert von 1 bedeutet, die Vektoren zeigen in dieselbe Richtung (maximale Ähnlichkeit).
* Ein Wert von 0 bedeutet, die Vektoren sind orthogonal (keine Ähnlichkeit).
* Ein Wert von -1 bedeutet, die Vektoren zeigen in entgegengesetzte Richtungen (maximale Unähnlichkeit).

````python
# Wir haben bereits die Embeddings für unsere Beispielsätze generiert.
# `embeddings_hf` enthält die Vektoren.

# Ähnlichkeit zwischen Satz 1 und Satz 2
similarity_hf_1_2 = cosine_similarity(
    embeddings_hf[0].reshape(1, -1),  # Ersten Vektor als Zeilenvektor
    embeddings_hf[1].reshape(1, -1)   # Zweiten Vektor als Zeilenvektor
)
print(f"Ähnlichkeit (HF) zwischen '{sentences_hf[0]}' UND '{sentences_hf[1]}': {similarity_hf_1_2[0][0]:.4f}")

# Ähnlichkeit zwischen Satz 1 und Satz 3
similarity_hf_1_3 = cosine_similarity(
    embeddings_hf[0].reshape(1, -1),
    embeddings_hf[2].reshape(1, -1)
)
print(f"Ähnlichkeit (HF) zwischen '{sentences_hf[0]}' UND '{sentences_hf[2]}': {similarity_hf_1_3[0][0]:.4f}")

# Ähnlichkeit zwischen Satz 1 und Satz 4
similarity_hf_1_4 = cosine_similarity(
    embeddings_hf[0].reshape(1, -1),
    embeddings_hf[3].reshape(1, -1)
)
print(f"Ähnlichkeit (HF) zwischen '{sentences_hf[0]}' UND '{sentences_hf[3]}': {similarity_hf_1_4[0][0]:.4f}")


print("\n**Aufgabe zum Ausprobieren:**") # Geändert zu print, da @markdown hier nicht funktioniert
print("Vergleiche andere Satzpaare und beobachte die Ähnlichkeitswerte. Macht das Ergebnis Sinn?")

# Beispiel: Ähnlichkeit aller Sätze mit dem ersten Satz
print("\n--- Ähnlichkeit aller Sätze mit dem ersten Satz (Hugging Face) ---")
for i in range(len(sentences_hf)):
    similarity = cosine_similarity(embeddings_hf[0].reshape(1, -1), embeddings_hf[i].reshape(1, -1))
    print(f"Ähnlichkeit von '{sentences_hf[0]}' mit '{sentences_hf[i]}': {similarity[0][0]:.4f}")
````

#### Hugging Face API Token (User Access Token)

Ein Hugging Face API Token (User Access Token) erlaubt dir:
* Zugriff auf private Modelle oder Datensätze (falls du die Berechtigung hast).
* Das Hochladen eigener Modelle.
* Manchmal höhere Ratenlimits bei der Nutzung der Inference API.
* Für die meisten `sentence-transformers` Modelle, die öffentlich verfügbar sind (wie das oben verwendete), ist **kein** Token für den Download und die Nutzung notwendig. Er wird aber für bestimmte API-Interaktionen oder private Ressourcen benötigt.

**Schritte zum Generieren eines Hugging Face API Tokens:**
1.  Gehe zu [huggingface.co](https://huggingface.co/) und erstelle einen Account oder logge dich ein.
2.  Klicke auf dein Profilbild oben rechts und wähle "Settings".
3.  Navigiere im linken Menü zu "Access Tokens".
4.  Klicke auf "New token".
5.  Gib dem Token einen Namen (z.B. "colab_embeddings_tutorial") und wähle eine Rolle (meistens ist "read" ausreichend, es sei denn, du möchtest Modelle hochladen, dann "write").
6.  Klicke auf "Generate a token".
7.  **Kopiere den Token sofort!** Er wird dir nur einmal angezeigt.

**API Token in Colab hinterlegen:**
1.  Klicke in der linken Seitenleiste von Colab auf das Schlüssel-Symbol (🔑, "Secrets").
2.  Klicke auf "+ Add a new secret".
3.  Gib als **Namen** für das Secret z.B. `HF_TOKEN` ein.
4.  Füge deinen kopierten Hugging Face API Token als **Wert** ein.
5.  Aktiviere den Schalter "Notebook access", damit dein Notebook auf dieses Secret zugreifen kann.
6.  Du kannst das Secret dann im Code so abrufen:

### (Optional) 🧪 6. Embeddings mit Google AI Studio (z.B. Gemini `embedding-001`)

Wenn du einen Google AI Studio API Key generiert und hinterlegt hast (siehe Schritt 5), kannst du auch die Embedding-Modelle von Google verwenden.
Das aktuelle Embedding-Modell von Google ist `models/embedding-001`.

**Wichtig:** Stelle sicher, dass du deinen `GOOGLE_API_KEY` in den Colab Secrets hinterlegt hast und den Schalter "Notebook access" aktiviert hast, bevor du diese Zelle ausführst!

````python
import google.generativeai as genai
import numpy as np # Sicherstellen, dass numpy importiert ist für np.array()
from google.colab import userdata # Sicherstellen, dass userdata importiert ist

# Versuche, den API Key zu laden und zu konfigurieren
GOOGLE_API_KEY = None
try:
    GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
except userdata.SecretNotFoundError:
    print("⚠️ Google API Key 'GOOGLE_API_KEY' nicht in Colab Secrets gefunden.")
    print("➡️ Bitte füge deinen Key hinzu (siehe Anleitung in Sektion 5) und führe diese Zelle erneut aus, um Google Embeddings zu testen.")
except Exception as e:
    print(f"Ein unerwarteter Fehler beim Laden des Google API Keys ist aufgetreten: {e}")

if GOOGLE_API_KEY:
    try:
        genai.configure(api_key=GOOGLE_API_KEY)
        print("✅ Google AI Studio API Key konfiguriert.")

        # Modell für Embeddings
        # https://ai.google.dev/models/gemini (siehe "Embeddings")
        google_embedding_model_name = "models/embedding-001"

        # Beispieltext
        text_to_embed_google = "Hallo von Google AI Studio!"
        print(f"\nGeneriere Google Embedding für: '{text_to_embed_google}'")

        # Embedding generieren
        # Task Types: RETRIEVAL_QUERY, RETRIEVAL_DOCUMENT, SEMANTIC_SIMILARITY, CLASSIFICATION, CLUSTERING
        result_google = genai.embed_content(
            model=google_embedding_model_name,
            content=text_to_embed_google,
            task_type="SEMANTIC_SIMILARITY"
        )
        google_embedding_vector = result_google['embedding']
        print(f"Dimension des Google Embeddings: {len(google_embedding_vector)}")
        print(f"Die ersten 10 Werte des Google Embeddings: {google_embedding_vector[:10]}")

        # Embeddings für mehrere Sätze
        sentences_google = [
            "Wie funktioniert Quantencomputing?",
            "Erkläre die Grundlagen der Quantenmechanik.",
            "Was ist das beste italienische Restaurant in Berlin?"
        ]
        print(f"\nGeneriere Google Embeddings für {len(sentences_google)} Sätze.")

        # Batch-Embedding ist effizienter
        result_google_batch = genai.embed_content(
            model=google_embedding_model_name,
            content=sentences_google,
            task_type="SEMANTIC_SIMILARITY"
        )
        google_embeddings_batch = result_google_batch['embedding'] # Dies ist eine Liste von Embedding-Vektoren

        print(f"Anzahl der generierten Google Embeddings: {len(google_embeddings_batch)}")
        if google_embeddings_batch and len(google_embeddings_batch) > 0:
            print(f"Dimension jedes Google Embeddings: {len(google_embeddings_batch[0])}")

        # Kosinus-Ähnlichkeit mit Google Embeddings
        if len(google_embeddings_batch) >= 2:
            # Konvertiere die Liste der Listen in ein NumPy Array für cosine_similarity
            google_embeddings_batch_np = np.array(google_embeddings_batch)

            similarity_google_0_1 = cosine_similarity(
                google_embeddings_batch_np[0].reshape(1, -1),
                google_embeddings_batch_np[1].reshape(1, -1)
            )
            print(f"\nÄhnlichkeit (Google) zwischen '{sentences_google[0]}' UND '{sentences_google[1]}': {similarity_google_0_1[0][0]:.4f}")

            similarity_google_0_2 = cosine_similarity(
                google_embeddings_batch_np[0].reshape(1, -1),
                google_embeddings_batch_np[2].reshape(1, -1)
            )
            print(f"Ähnlichkeit (Google) zwischen '{sentences_google[0]}' UND '{sentences_google[2]}': {similarity_google_0_2[0][0]:.4f}")

    except Exception as e:
        print(f"Ein Fehler ist bei der Verwendung der Google AI API aufgetreten: {e}")
        print("Stelle sicher, dass die 'google-generativeai' Bibliothek korrekt installiert ist, der API Key gültig ist und dein Google Cloud Projekt für die API aktiviert ist.")
else:
    # Überprüfen, ob der Key nicht gefunden wurde oder einfach nicht gesetzt ist
    key_not_found = False
    try:
        # Dieser Aufruf dient nur dazu, SecretNotFoundError zu fangen, falls der Key nicht existiert.
        # Der eigentliche Wert wird nicht verwendet, wenn GOOGLE_API_KEY bereits None ist.
        if GOOGLE_API_KEY is None:
            userdata.get('GOOGLE_API_KEY')
    except userdata.SecretNotFoundError:
        key_not_found = True
    except Exception:
        pass # Andere Fehler ignorieren, da wir nur an SecretNotFoundError interessiert sind

    if not key_not_found and GOOGLE_API_KEY is None:
         print("ℹ️ Überspringe Google AI Studio Embedding-Beispiel, da der API Key nicht konfiguriert wurde oder nicht zugänglich ist.")
````
